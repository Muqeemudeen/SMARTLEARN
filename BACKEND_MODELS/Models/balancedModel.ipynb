```python
import pandas as pd
import numpy as np
import random
import os

# Number of records to generate
num_records = 2000

# Generate CGPA values between 4 and 10
cgpas = np.random.uniform(4, 10, num_records)

# Assign domains based on CGPA
assigned_domains = []
for cgpa in cgpas:
    if cgpa < 7:
        assigned_domains.append("Web Development")
    else:
        # Randomly assign "Data Science" or "Cloud Computing" for CGPA >= 7
        assigned_domains.append(random.choice(["Data Science", "Cloud Computing"]))

# Generate random study hours between 1 and 5 hours per day
study_hours = np.random.randint(1, 6, num_records)

# Generate random degrees, ensure variety
degrees = ["IT", "CSE", "IOT", "Other"]
assigned_degrees = [random.choice(degrees) for _ in range(num_records)]

# Randomly choose whether the user revisits the course material daily or weekly
revisit_frequencies = ["Daily", "Weekly"]
assigned_revisit = [random.choice(revisit_frequencies) for _ in range(num_records)]

# Create a DataFrame from the generated data
data = {
    "CGPA": cgpas,
    "Domain": assigned_domains,
    "Hours Spent": study_hours,
    "Degree": assigned_degrees,
    "Revisit Frequency": assigned_revisit
}

df = pd.DataFrame(data)

# Get the current working directory to save the file in the same location
current_directory = os.getcwd()

# Define the path for saving the CSV file
file_path = os.path.join(current_directory, "balanced_domain_recommendation_data.csv")

# Save the dataset to a CSV file in the current directory
df.to_csv(file_path, index=False)

# Display the first few rows of the generated data
print(df.head())

# Optionally, check the distribution of domains
print(df['Domain'].value_counts())

# Confirm file saved
print(f"✅ Dataset saved as '{file_path}'")
```
